```python
is_null_frame = pre_data.isnull().mean()*100
```

```python
is_null_frame.iloc[0]
```

```python
len(is_null_frame)
```

```python
num = 0
ls_nan = []
for i in range(len(is_null_frame)):  
    if is_null_frame.iloc[i] > 0.9:  # 1차원 Series이므로 iloc[i]만 사용
        print(is_null_frame.index[i])  # 컬럼명 출력
        ls_nan.append(is_null_frame.index[i])
        num=num+1
print(num)
```

```python
print(len(ls_nan))
print(pre_data.shape)
```

```python
for name in ls_nan:
    pre_data = pre_data.drop(columns=[name])
```
